```python
#Q8
l = [10, 40, 20, 30, 50] #solution other than l[-2]
largest = 0
second = 0

for num in l:
    if num > largest:
        second = largest
        largest = num
    elif num > second and num != largest:
        second = num
print("Scond largest -", second)            
# Number Largest	Second
# 10	 10	        10
# 40	 40	        10
# 20	 40	        20
# 30	 40	        30
# 50	 50	        40
```

```python
#Q5
l = [2,7,11,15]
target = 9
for i in range(len(l)):
    for j in range(i+1 ,len(l)):
        if(l[i] + l[j]) == target:
            print("pair =",(l[i],l[j]))
            break 
```

```python
#Q7
l = [1,2,2,2,3,3,3]
freq = {}
for num in l:
    if num in freq:
        freq[num] +=1
    else:
        freq[num] = 1
for num , count in freq.items():
    print(num , "->" , count) 
        


```
