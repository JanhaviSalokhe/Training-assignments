```python
#Q8
l = [10, 40, 20, 30, 50] #solution other than l[-2]
largest = 0
second = 0

for num in l:
    if num > largest:
        second = largest
        largest = num
    elif num > second and num != largest:
        second = num
print("Scond largest -", second)            
# Number Largest	Second
# 10	 10	        10
# 40	 40	        10
# 20	 40	        20
# 30	 40	        30
# 50	 50	        40
```
