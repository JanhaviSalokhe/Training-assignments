### Lambda function

#### Basic questions

```python
# 1. Write a Python lambda function to return the square of a number.
re = (lambda x : x**2)
print(re(9))
```

```python
# 2. Write a Python lambda function to return the cube of a number.
re = (lambda x : x**3)
print(re(4))
```

```python
#   3. Write a Python lambda function to add two numbers.
re = lambda x , y : x+y
print(re(15,25))
```

```python
# 4. Write a Python lambda function to return the greater of two numbers
re = lambda a,b : a if a>b  else  b
print(re(18,27))
```

```python
# 5. Write a Python lambda function to check whether a number is even.
re = lambda x :  'True' if x%2 == 0 else 'False'
print(re(14))
```

```python
# 6. Write a Python lambda function to return Positive, Negative, or Zero based on a number.
re = lambda x : 'Positive' if x > 0  else 'negative' if x <0 else 'Zero'
# nested
print(re(-8))
```

```python
# 7. Write a Python lambda function to return the first character of a string.
re = lambda x : x[0]
print(re("Python"))
```

```python
# 8. Write a Python lambda function to return the length of a string
re = lambda x : len(x)
print(re("LihaTech"))
```

```python
# 9. Write a Python lambda function to convert a string to uppercase
re = lambda x : x.upper
print(re("python"))
```

```python
# 10. Write a Python lambda function to multiply three numbers.
re = lambda a,b,c : a*b*c
re(2,3,4)
```

```python
# 11. Write a Python program to sort a list of tuples by the second value using a lambda function.
items = [(1,30),(2,10),(3,20)]
re = sorted(items ,key = lambda i : i[1])
re
```

```python
# 12. Write a Python program to sort words by their length using a lambda function.
words = ['python', 'AI', 'data', 'cloud']
re = sorted(words , key = lambda x : len(x) )
re
```

```python
# 13. Write a Python program to square all numbers in a list using map() and lambda.
numbers = [1, 2, 3, 4, 5]
x = list(map(lambda x : x**2 , numbers))
print(x)
```

```python
# 14. Write a Python program to select all even numbers using filter() and lambda.
numbers = [1, 2, 3, 4, 5, 6]
x = list(filter(lambda a :  a %2 ==0  , numbers))
print(x)
```

```python
# 15. Write a Python program to select words beginning with A using filter() and lambda.
words = ['Apple', 'Python', 'AI', 'Data', 'Agent']
x = list(filter(lambda x : x[0] == 'A' , words))
print(x)
```

```python
# 16. Write a Python program to convert Celsius values to Fahrenheit using map() and lambda.
celsius = [0, 20, 30, 40]
x = list(map(lambda a : (a*1.8) + 32 , celsius) )
x
```

```python
# 3. Write a Python program to apply a 10% discount to every price using map() and lambda.
prices = [500, 1000, 1500, 2000]
re = list(map(lambda x : x - (x * 0.1) , prices))
re
```

```python
# 4. Write a Python program to filter student records with marks of 75 or above using filter() and lambda.
students = [('Asha', 82), ('Ravi', 68), ('Neha', 91), ('Aman', 75)]
re = list(filter(lambda x : x[1] >= 75 , students))
re
```

```python
# 5. Write a Python program to calculate the product of all numbers using reduce() and lambda.
from functools import reduce
num = [2,3,4,5]
re = reduce(lambda x ,y : x * y , num)
print("Product =" , re)
```

```python
# 9. Write a Python program to find the dictionary key having the maximum value using a lambda key.
scores = {'Asha': 82, 'Ravi': 74, 'Neha': 91}
re = max(scores , key = lambda x : scores[x])
re
```

```python
# 10. Write a Python program to sort student records by marks in descending order and then by name in
# ascending order using lambda.
students = [('Ravi', 85), ('Asha', 92), ('Neha', 85), ('Aman', 92)]
re = sorted(students ,key  = lambda x : (-x[1], x[0]))
re
```

```python
# 2. Write a Python lambda expression to calculate tax using slabs: 0% up to 300000, 5% up to 600000, and
# 10% above 600000.

income = 800000
re = (lambda x : x if x <=  300000 else x - (x* 5/100) if x >=  600000 else x - ( x* 10/100) )(income)
re

```

```python
# 3. Write a Python program using filter(), map(), and lambda to return the squares of only even numbers.
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even_squares = list(filter(lambda x :  x%2 == 0 ,numbers))
squares = list(map(lambda  x : x**2 , even_squares))
squares
```

```python
# 4. Write a Python program to sort file-version tuples by major, minor, and patch numbers using a lambda
# key.
versions = [('app', '1.2.10'), ('app', '1.10.2'), ('app', '1.2.3')]
re = sorted(versions , key = lambda x: x[1] , reverse = True)
re
```

```python
# 5. Write a Python program to sort points by their Manhattan distance from the origin using a lambda key.
points = [(3, 4), (1, 1), (-2, 2), (0, 5)]
re = sorted(points, key = lambda x : abs(x[0]) + abs(x[1]))
re
```

```python
# 6. Write a Python program to sort ISO date strings chronologically using a lambda key that converts date
# parts to integers.

dates = ['2026-12-01', '2025-05-10', '2026-01-15']
re = sorted(dates, key = lambda date : tuple(map(int, date.split('-'))))
re
```

```python
# 7. Write a Python program to find the longest word using max() with a lambda key.
words = ['AI', 'Python', 'MachineLearning', 'Data']
re = max(words , key = lambda x : len(x))
re
```
