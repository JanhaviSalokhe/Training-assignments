```python
#9. Write a Python program to count the frequency of characters in a string using a dictionary.
d =  "aabcc"
freq = {}
for ch in d:
    if ch in freq:
        freq[ch] +=1
    else:
        freq[ch] = 1
for ch , freq in freq.items():
    print(ch,":" ,freq , end = " ")            
```
