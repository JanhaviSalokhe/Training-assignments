### List comprehension

#### Basic questions

```python
#1. Write a Python program to create a list of squares using list comprehension.
number = [1,2,3,4,5]
x = [i**2 for i in number ]
print(x)
```

```python
#2. Write a Python program to extract all even numbers from a list using list comprehension.
numbers = [1,2,3,4,5,6,7,8]
x = [i for i in numbers if i%2 == 0]
print(x)
```

```python
# 3. Write a Python program to extract all odd numbers from a list using list comprehension.
numbers = [1,2,3,4,5,6,7,8]
x = [i for i in numbers if i%2 != 0]
print(x)
```

```python
# 4. Write a Python program to convert all words in a list to uppercase using list comprehension.
words = ['python', 'java', 'sql']
x = [i.upper() for i in words ]
print(x)
```

```python
# 5. Write a Python program to create a list containing the length of each word.
words = ['AI', 'Python', 'LihaTech']
x = [len(i) for i in words]
print(x)
```

```python
# 6. Write a Python program to select words having more than four characters.
words = ['data', 'python', 'AI', 'model', 'cloud']
x = [i for i in words if len(i) > 4]
print(x)
```

```python
# 7. Write a Python program to double every number in a list using list comprehension.
numbers = [2, 4, 6, 8]
x = [ i*2 for i in numbers]
print(x)
```

```python
# 8. Write a Python program to extract only positive numbers from a list.
numbers = [-5, 3, -1, 0, 7, -2, 9]
x = [i for i in numbers if i>0]
print("Positive numbers",x)
```

```python
# 9. Write a Python program to convert a string into a list of its characters using list comprehension.
text = "Python"
x = [i for i in text]
print(x)
```

```python
# 10. Write a Python program to extract only vowels from a string.
text = "LIhaTech Kolhapur"
x = [i.lower() for i in text if i.lower() in 'aeiou']
print(x)
```

```python
# 11. Write a Python program to replace every negative number in a list with zero.
numbers = [4, -2, 7, -5, 3]
x = [0 if i < 0 else i for i in numbers]
print(x)
```

```python
#12. Write a Python program to create a list containing the first character of each word.
Input_words = ['Python', 'Machine', 'Learning']
x = [i[0] for i in Input_words]
print(x)
```

```python
#13. Write a Python program to create pairs of each number and its square.
numbers = [1, 2, 3, 4]
x = [ (i, i**2) for i in numbers]
print(x)
```

```python
#14. Write a Python program to flatten a two-dimensional list using list comprehension.
matrix = [[1, 2], [3, 4], [5, 6]]
x = [i for row in matrix  for i in row]
print(x)
# for row in matrix: First, loop through the main list to get each sublist (e.g., [1, 2]).for num in row: Next, loop through that sublist 
# to get the individual numbers (e.g., 1, then 2).num: Finally, place the individual number at the very front to add it to the new list
```

```python
#15. Write a Python program to generate all multiples of 3 from 1 to 20 using list comprehension.
# Range = 1 to 20
x = [ i for i in range(1,20) if i%3 == 0 ]
print(x)
```

```python
#16. Write a Python program to convert a list of numeric strings into integers.
values = ['10', '20', '30', '40']
x = [int(i) for i in values]
print(x)
```

### Intermediate questions

```python
# 1. Write a Python program to label each number as Even or Odd using conditional list comprehension.
numbers = [1, 2, 3, 4, 5]
x = ['even' if i%2 == 0 else 'odd' for i in numbers]
print(x)
```

```python
# 2. Write a Python program to transpose a matrix using nested list comprehension.
matrix = [[1, 2, 3], [4, 5, 6]]
# x = zip(*matrix)-------------------------
print(x)
```

```python
# 3. Write a Python program to find common elements between two lists using list comprehension
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

x = [i for i in list1 if i in list2]
print(x)
```

```python
# 4. Write a Python program to remove empty or whitespace-only strings from a list.
values = ['Python', '', ' ', 'AI', 'Data']
x = [i for i in values if i is i.isalpha()]
print(x)
```

```python
# 5. Write a Python program to flatten a nested list but keep only even numbers.
matrix = [[1, 2, 3], [4, 5, 6], [7, 8]]
x = [i for row in matrix for i in row if i %2 == 0]
print(x)
```

```python
# 6. Write a Python program to extract the names of students whose score is 75 or above.
students = [('Asha', 82), ('Ravi', 68), ('Neha', 91), ('Aman', 75)]
x = [i[0] for i in students if i[1] >= 75 ]
print(x)
```

```python
# 7. Write a Python program to generate the Cartesian product of two lists.
list1 = [1, 2]
list2 = ['A', 'B']
x = [(i,j) for i in list1 for j in list2]
print(x)
```

```python
# 8. Write a Python program to generate all prime numbers between 2 and 30 using list comprehension.
x = [i for i in range(2,30)  if all(i%n != 0 for n in range(2,i)) ]
print(x)
```

```python
# 9. Write a Python program to extract all palindrome words from a list.
words = ['level', 'python', 'madam', 'data', 'radar']
x = [i for  i in words if i == i[::-1]]
print(x)
```

```python
# 10. Write a Python program to extract the main diagonal elements of a square matrix.
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
x = [matrix[i][i] for i in range(len(matrix))]
print(x)
```

### ADVANCED QUESTIONS

```python
#1. Write a Python program to find all unique elements that occur more than once while preserving their first
# appearance order.
numbers = [4, 2, 4, 3, 2, 5, 4, 3]
x = list(dict.fromkeys([i for i in numbers if numbers.count(i) > 1]))
x
```

```python
# 2. Write a Python program to generate all pairs whose sum is equal to a given target.
numbers = [1, 2, 3, 4, 5, 6]
target = 7
x = [(i,j) for i in numbers for j in numbers if i + j == target]
x
```

```python
# 3. Write a Python program to create the transpose of a 3 x 3 matrix using nested list comprehension.
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

```

```python
# 4. Write a Python program to flatten a three-level nested list into a single list.
data = [[[1,2],[3]],[[4,5],[6]]]
x = [i for cells in data  for subcells in cells  for i in subcells]
x
```

```python
# 5. Write a Python program to calculate the sum of each row of a matrix using list comprehension.
matrix = [[2, 4, 6], [1, 3, 5], [10, 20, 30]]
x = [sum(row) for row in matrix]
x
```

```python
# 6. Write a Python program to assign grades to marks using a conditional list comprehension: A for 90+, B
# # for 75-89, C for 60-74, and D below 60.

marks = [95, 82, 67, 54, 90]
x = ['A' if i >= 90 else 'B' if i >= 75 else 'C' if i >= 60 else 'D' for i in marks]
x
```

```python
# 7. Write a Python program to convert comma-separated student records into tuples.
records = ['Asha,85', 'Ravi,72', 'Neha,91']
x = [tuple(i.split(',')) for i in records]
x
```

```python
# 8. Write a Python program to create a multiplication table from 1 to 5 as a nested list.
x = [[i * n for i in range(1,6)] for n in range(1,6)]
x
```

```python
# 9. Write a Python program to generate all unique Pythagorean triples where 1 <= a < b < c <= 20.
x = [(a,b,c) for a in range(1,21) for b in range(a,21) for c in range(b,21) if a**2 + b**2 == c**2]
x
```

```python
# 10. Write a Python program to generate sliding windows of size 3 from a list using list comprehension.
numbers = [10, 20, 30, 40, 50, 60]
x = [numbers[i:i+3] for i in range(len(numbers)-2)]
print(x)
```
