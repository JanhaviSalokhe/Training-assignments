```python
#Q16
s = input()
upper = 0
lower = 0
digit = 0
special = 0

for ch in s:
    if ch.islower():
        lower +=1
    elif ch.isupper():
        upper +=1
    elif ch.isdigit():
        digit +=1  
    else:
        special +=1
print("uppercase",upper)
print("lowercase",lower)
print("digit",digit)
print("Special",special)            

```
