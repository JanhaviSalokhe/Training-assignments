```python
#Q2
t = (1,2,3)
t2 = (4,5,6)
result = []
for i in range (len(t)):
   result.append(t[i] + t2[i])
print(tuple(result))   
```

```python
#Q3
t = (1,2,2,3,4,4,4)
freq = {}
result = []
for i in t:
    if i in freq:
        freq[i] +=1
    else:
        freq[i] = 1
for i , freq  in freq.items():
    if freq >= 2 :
        result.append(i)
print("Repeated values",tuple(result))        



```
