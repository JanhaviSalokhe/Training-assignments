```python
# 1.  Write a Python function that accepts a name and returns a greeting message.
def greet(name):
   return f'Hello, {name}!'

n = input()
print(greet(n))
```

```python
# 2.  Write a Python function that accepts two numbers and returns their sum.
def add(a,b):
    add = a+b
    return add
print(add(12,8))    
```

```python
# 3.  Write a Python function that returns the greater of two numbers
def greater(a,b):
    if a >b:
        return a
    else:
        return b


a = 35
b = 48
print(greater(a,b))        
```

```python
# 4.  Write a Python function that checks whether a number is even or odd.
def check(n):
    if n%2 == 0:
        return f'{n} is even number'
    else:
        return f'{n} is odd number'

n = 26
print(check(n))            
```

```python
# 5.  Write a Python function that returns the square of a number.
def square(n):
    return n*n

n = 7
print(f' Square = {square(n)}')   
```

```python
# 6.  Write a Python function to calculate the area of a rectangle.
def rectangle(l,b):
    return l*b

l = 12
b = 5
print("Area =",rectangle(l,b)) 
```

```python
# 7.  Write a Python function to convert Celsius temperature to Fahrenheit.
# (9°C × 9/5) + 32 = 48.2°F
def conversion(x):
    return (x* 9/5)+32

x = 25
print("Fahrenheit =",conversion(x))    
```
