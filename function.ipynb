```python
# 5.  Write a Python function to count the frequency of each element in a list and return a dictionary.
def fre(items):
    freq = {}
    for i in items:
        if i in freq:
            freq[i] +=1
        else:
            freq[i] = 1 
    return freq        

items = ['A', 'B', 'A', 'C', 'B', 'A']
print(fre(items))
```

```python
# 7.  Write a recursive Python function to return the nth Fibonacci number.
def fibonaaci(n):
    sum = 0
    for i in range(0,n-1):
        sum +=i
    return sum

n = 8
print(fibonaaci(n))
    
```

```python
# 8.  Write a Python function to calculate income tax using slabs: 0% up to 300000, 5% for the next 300000,
# and 10% for the remaining amount.
def income_tax(income):
    if income <= 300000:
        tax = 0
    elif income <= 600000:
        tax =  (income-300000)* 0.05
    else :
        tax =  300000 * 0.05 + (income - 600000)* 0.10
    return tax    
# ₹8,00,000 - ₹6,00,000 = ₹2,00,000
income = 800000
print("Tax =",income_tax(income))

       
```

```python
# 9.  Write a Python function containing a nested helper function to calculate gross salary from basic salary,
# HRA, and DA percentages.
def main_function(basic , HRA , DA):

    def helper_function(percent):
        return basic * percent/100

    hre = helper_function(HRA) 
    de = helper_function(DA)

    return basic + hre + de

basic = 30000
HRA = 20
DA = 10

print("Gross salary =",main_function(basic , HRA,DA))
    
```

```python
# 10.  Write a Python function to validate a password and return both status and reason. Require at least 8
# characters, one uppercase letter, and one digit.
def password(s):
    if len(s) >= 8:
        if any(char.isupper() for char in s):
            if any(char.isdigit() for char in s):
                print("Valid = True \nReason = Password accepted")
    else:
        print("Invalid")            

            
s = "Python123"
password(s)   

```

```python
# 6.  Write a Python function to return the k most frequent elements in a list.
def duplicate(nums):
    list = []
    dupli = []
    for i in nums:
        if i not in list:
            list.append(i)
        else:
            dupli.append(i)
    return dupli

numbers = [1,1,1,2,2,3]
print(duplicate(numbers))     

```
