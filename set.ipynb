```python
# 15.  Write a Python program to create a frozen set and demonstrate its usage.
s = [1, 2, 3, 4]
frozen = frozenset(s)
print(frozen)
#frozenset is immutable
```

```python
# 16.  Write a Python program to find common unique words from two sentences using sets.
sentence1 = "python is easy and powerful"
sentence2 = "python is popular and useful"
sen1_set = sentence1.split()
sen2_set = sentence2.split()
unique = set(sen1_set).intersection(set(sen2_set))
print(unique)

```

```python
# 1.  Write a Python program to find unique vowels present in a sentence using a set.
s = "Education is a powerful tool"
s.lower()
vowels = 'aeiou'
listt = []
for i in s:
    if i in vowels:
        listt.append(i)
print(set(listt))        
```

```python
# 2.  Write a Python program to find the common elements among three sets.
A = {1, 2, 3, 4}
B = {2, 3, 4, 5}
C = {3, 4, 6}
pr = A.intersection(B).intersection(C)
print(pr)
```

```python
# 3.  Write a Python program to find elements that belong to exactly one of two sets.
A = {1, 2, 3}
B = {3, 4, 5}
J = A.symmetric_difference(B)
print(J)
```

```python
# 4.  Write a Python program to find missing numbers from 1 to n using set operations.
S = {1, 2, 4, 5, 7, 10}
n = 10
listt = []
for i in range(1,n):
    if i not in S:
        listt.append(i)
print(set(listt))
```

```python
# 5.  Write a Python program to calculate the Jaccard similarity of two sets.
# jaccard similarity means - A intersection B/ A uniion B

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
intersection_set = a & b
union_set = a | b
jaccard = len(intersection_set)/len(union_set)
print(jaccard)

```

```python
# 6.  Write a Python program to find duplicate words in a sentence using sets.
s = "python is easy and python is powerful"
splitt = s.split()
unique = []
duplicate = []
for i in splitt:
    if i not in unique:
        unique.append(i)
    else:
        duplicate.append(i)
print("Duplicate word =",set(duplicate))      
 

```

```python
# 9. Write a Python program to create a set containing squares of all even numbers from 1 to 10.
x1 = set(map(lambda x : x*x , filter(lambda x:  x%2 == 0 , range(1,11))))
print(x1)
```

```python
# 10.  Write a Python program to find characters that are common to every word in a list.
Words = ['bella', 'label', 'roller']
first_word = set(Words[0])

for word in Words[1:]:
    common = first_word.intersection(set(word))
print(" Common Characters =", common)    



```
